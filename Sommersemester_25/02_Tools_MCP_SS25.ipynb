# Kapitel 3 MCP Server

Ein Model Context Protocol Server (MCP-Server) ist ein eigenständiger Dienst, der es einer KI (z. B. einem Sprachmodell wie Claude oder Llama) auf standardisierte Weise erlaubt, auf externe Datenquellen, APIs oder Anwendungsfunktionen zuzugreifen.

Man kann sich MCP wie einen universellen Anschluss für KI-Anwendungen vorstellen, ähnlich wie USB-C bei Laptops:

Statt für jedes Gerät (Datenbank, Kalender, Maps, Aktienkurse…) eine eigene, komplexe Verbindung zu bauen, gibt es einheitliche Steckplätze, über die die KI mit der Außenwelt reden kann.

Das bedeutet:

- Statt für jeden neuen Dienst individuell **„Glue Code“** zu schreiben, kann man auf **einheitliche Funktionsaufrufe** zurückgreifen.

- Die Kommunikation mit dem MCP-Server erfolgt über ein **offenes, standardisiertes JSON-Format**, unabhängig vom zugrundeliegenden Anbieter (Google, Microsoft, Open Source usw.).

- Der Server übernimmt die Rolle einer **Middleware** zwischen dem KI-Modell und der realen Welt: Er bündelt externe Funktionen, Daten und Services in einer strukturierten, KI-freundlichen Form.

Ziel ist es, interne wie externe Systeme **herstellerunabhängig** und **effizient** mit Sprachmodellen verbinden zu können. Egal ob es sich um eine lokale Datei, eine Websuche oder eine komplexe Drittanbieter-API handelt.

Mehr zur Architektur und Vorteielen des MCP Protokolls kannst du hier nachlesen: https://k33g.hashnode.dev/understanding-the-model-context-protocol-mcp

## 3.1 Funktionalitäten eines MCP Server

Ein MCP-Server zeichnet sich im Wesentlichen durch 3 Hauptfunktionen aus. Diese werden standardisiert über bestimmte Methoden zur Verfügung gestellt.

Finde in folgender Dokumentation heraus, welche Bestandteile einen MCP Server ausmachen und wofür diese verwendet werden: https://github.com/modelcontextprotocol/python-sdk/blob/main/README.md#what-is-mcp

Ein MCP-Server kann dem Modell im Wesentlichen drei Arten von Zugriffsmöglichkeiten bereitstellen. Diese werden standardisiert über bestimmte Methoden zur Verfügung gestellt:

**1. Tools:**

Dies sind aktive Funktionen, die das Modell ausführen lassen kann. Tools sind vergleichbar mit Endpunkten in klassischen APIs, aber KI-kompatibel und automatisch beschreibbar.

Beispiel: get_route, get_weather, buy_stock



**2. Resources:**

Dies sind statische oder abrufbare Datenquellen, auf die das Modell zugreifen kann. Sie dienen der Erweiterung des Kontexts oder zur Verwendung in Kombination mit Tool-Calls.

Beispiel: Zugriff auf einen Firmendatensatz, Dokumentenspeicher, Tabellen



**3. Prompts:**

Dies sind vordefinierte Eingabehilfen oder Vorlagen, die das Verhalten des Modells steuern. Sie helfen, die Modellantworten kontrollierbar und kontextangepasst zu gestalten.

Beispiel: „Formuliere E-Mails höflich“, „Fasse Suchergebnisse strukturiert zusammen“



Alle drei Kategorien können über standardisierte Methoden abgefragt werden:

- `list_tools`

- `list_resources`

- `list_prompts`

Das macht es möglich, dass ein Modell dynamisch neue Fähigkeiten und Inhalte „entdecken“ kann – ohne dass ein Entwickler explizit wissen muss, was ein Server genau anbietet.
