# Blender mit MCP‑Server nutzen

Dieses Notebook erweitert Kapitel 2 (Model Context Protocol) und zeigt, wie Blender über ein Sprachmodell gesteuert wird. Zuerst wird der MCP‑Server in Blender eingerichtet, anschließend über Cursor verbunden, sodass Szenen via Prompts editiert werden können.

## Kapitel 1: Einrichtung

**Offizielle Anleitung**  
<https://github.com/ahujasid/blender-mcp/tree/main>

### 1.1 Software installieren

| Programm | Download | Installationshinweise |
|----------|----------|-----------------------|
| Blender  | <https://www.blender.org/download/> | Standardoptionen |
| Cursor   | <https://www.cursor.com/downloads> | Standardoptionen |

*Medienlabor F01.303*  
Blender: `C:\Program Files\Blender Foundation\Blender 4.3\blender.exe`  
Cursor: `C:\Tools\cursor\Cursor.exe`

### 1.2 MCP‑Add‑on in Blender aktivieren

1. Blender starten → **Edit ▸ Preferences ▸ Add‑ons**  
2. Rechts oben ⋯ → **Install from Disk**  
3. Datei `blender_mcp/addon.py` wählen (liegt neben diesem Notebook) – alternativ herunterladen: <https://github.com/ahujasid/blender-mcp/blob/main/addon.py>  
4. Blender neu starten.  
5. Seitenleiste ( > ) öffnen → **Blender MCP** aktivieren.  
6. **Set Free Trial API Key** auswählen.  
7. **Connect to MCP Server** drücken; Port `9876` muss angezeigt werden.

### 1.3 uv installieren (Windows)

```powershell
irm https://astral.sh/uv/install.ps1 | iex
$env:Path = "C:\Users\<User>\.local\bin;$env:Path"
```

### 1.4 MCP‑Server in Cursor registrieren

1. Cursor öffnen → Zahnrad oben rechts → **MCP Server**  
2. **+ Add New Global MCP Server** wählen und einfügen:

```json
{
  "mcpServers": {
    "blender": {
      "command": "cmd",
      "args": ["/c", "uvx", "blender-mcp"]
    }
  }
}
```

3. Speichern. Grüner Statuspunkt bestätigt die Verbindung.

## Kapitel 2 – Sprachmodell einrichten

Cursor bietet drei Möglichkeiten, ein Sprachmodell zu nutzen:

### 1. Cursor-Konto
* Neues Konto enthält ein Startkontingent kostenloser Prompts.  
* Zur Auswahl stehen überwiegend Claude-Modelle.  
* Modellauswahl: rechter Rand des Chatfensters.

![Cursor-Modelle](ressource/cursor_models.png)

### 2. Externen Provider verbinden
* API-Schlüssel unter **⚙ Einstellungen → Models** hinterlegen.

![Custom Provider](ressource/cursor_custom_provider.png)

### 3. Ollama einbinden *(Admin-Rechte erforderlich; auf Medienlabor-Rechnern nicht verfügbar)*
* Funktioniert nur mit einem seperaten Proxy
* Anleitung: <https://medium.com/@hyperfox_/run-cursor-ai-for-free-with-open-source-llm-55396c1411b1>


## Kapitel 3 – Blender mit Sprachmodell steuern

### Voraussetzungen
* MCP-Server in Blender läuft  
* Verbindung zwischen Cursor und Blender aktiv

### Ablauf
1. In Blender alle Objekte löschen  
2. Im Cursor-Chat eingeben: `Create a tree in blender`  
3. Alle Tool-Calls bestätigen  
   * Dauerhaft automatisieren: **Settings → Features → Chat → Enable auto-run mode**

![Tool-Calls](ressource/cursor_tool_calls.png)

Ergebnis: Baum erscheint in der Blender-Szene.
